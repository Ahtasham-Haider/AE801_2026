## **Mathematical Formulation of the Blasius Boundary Layer Equation**

For a steady, incompressible flow of a uniform stream $U_\infty$ over a flat plate, applying the similarity transformation:
$$\eta = y\sqrt{\frac{U_\infty}{\nu x}}, \quad \frac{u}{U_\infty} = f'(\eta)$$

reduces the governing boundary layer equations to the third-order nonlinear ordinary differential equation known as the **Blasius equation**:
$$f''' + \frac{1}{2}ff'' = 0, \quad 0 < \eta < \infty \quad \text{--- (1)}$$

This equation is subject to the following boundary conditions:
* **At the wall ($\eta = 0$):** $f(0) = 0$ and $f'(0) = 0$ *(no-slip)*
* **In the free stream ($\eta \to \infty$):** $f'(\eta) \to 1$ *(free-stream)*


## **Reduction to a First-Order System**

To solve this third-order boundary value problem numerically, it is converted into a system of three coupled first-order ordinary differential equations. We define the state variables as follows:
$$Y_1 = f$$
$$Y_2 = Y_1' = f'$$
$$Y_3 = Y_2' = f''$$

Substituting these definitions into Equation (1) yields:
$$Y_3' + \frac{1}{2}Y_1Y_3 = 0$$

Rearranging and combining with the state definitions gives the complete system of first-order equations:
$$Y_1' = Y_2 \quad \text{--- (2)}$$
$$Y_2' = Y_3 \quad \text{--- (3)}$$
$$Y_3' = -\frac{1}{2}Y_1Y_3 \quad \text{--- (4)}$$

---

#### **Matrix Representation**

This system can be expressed in vector-matrix form as:
$$\mathbf{Y}' = \mathbf{G}(\mathbf{Y})$$

where
$$\mathbf{Y}' = \begin{bmatrix} Y_1' \\ Y_2' \\ Y_3' \end{bmatrix}, \quad \mathbf{G}(\mathbf{Y}) = \begin{bmatrix} Y_2 \\ Y_3 \\ -\frac{1}{2}Y_1Y_3 \end{bmatrix}$$

#### **Standard RK3 Formula**

Given a step size h (Δη) and a current point $(t_n, y_n)$, the next value $y_{n+1}$ is computed as:

$$\mathbf{y_{n+1}} = \mathbf{y_n} + \mathbf{\frac{h}{6}(k_1 + 4k_2 + k_3)}$$

Where the three slope increments (k-values) are defined as:

*   $k1 = f(t_n, y_n)$
*   $k2 = f(t_n + \frac{h}{2}, y_n + \frac{h}{2}k_1)$
*   $k3 = f(t_n + h, y_n - hk_1 + 2hk_2)$


```python
# Runge-Kutta 3 Function
def RK_3(Y, η_span, func, Δη):

    

    (ηi, ηf) = η_span       # Starting and End Point

    η_matrix = np.arange(ηi, ηf+Δη/2, Δη)               # It will hold time steps
    Y_matrix = np.zeros((len(η_matrix), len(Y)))        # It will hold f, f' f'' at different time steps      
    η_matrix[0] = ηi        # Initializing time steps  
    Y_matrix[0] = Y         # Intial Values of f, f', f''

    for i in range(len(η_matrix) - 1):
        k1 = Δη*func(η_matrix[i], Y_matrix[i])
        k2 = Δη*func(η_matrix[i] + Δη/2, Y_matrix[i] + k1/2)
        k3 = Δη*func(η_matrix[i] + Δη, Y_matrix[i] - k1 + 2*k2)
        Y_matrix[i+1] = Y_matrix[i] + (k1 + 4*k2 + k3)/6
    return η_matrix, Y_matrix
```

```python
# Shooting Method
def bisection_method(f, S, tol=1e-5, max_iter=1000):

    xL, xR = S               # Initial Guesses
    f = f                       # Function
    tol = tol                   # Tolerance
    max_iter = max_iter         # Maximum Iteration

    i = 0                       # Iteration
    err = abs(xR - xL)          # Initial Error

    if f(xL)*f(xR) > 0:
        raise ValueError("f(xL) and f(xR) must have opposite signs")

    while tol < err and i < max_iter:
        if f(xL) == 0:                  # if True xL is the root
            return xL
        elif f(xR) == 0:                # if True xR is the root
            return xR
        else:
            xM = (xL + xR)/2
            if f(xM) == 0:              # if True xM is the root
                return xM
            elif f(xL)*f(xM) < 0:       # Means f(xM) and f(xR) has same sign
                xR = xM
            else:
                xL = xM
        err = abs(xR - xL)
        i+=1

    print(f"Error = {err}")
    print(f"Iteration {i} completed !, \n")
    return (xL + xR)/2
```
